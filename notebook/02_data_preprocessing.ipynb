```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))

print("Project root:", PROJECT_ROOT)
```

```python
import pandas as pd
import numpy as np

from customer_intelligence_churn_prediction_system.data.load_data import (
    load_raw_data
)

from customer_intelligence_churn_prediction_system.preprocessing.clean_data import (
    clean_customer_data
)
```

```python
raw_df = load_raw_data()

print("Raw dataset shape:", raw_df.shape)
display(raw_df.head())
```

```python
cleaned_df = clean_customer_data(raw_df)

print("Raw shape:", raw_df.shape)
print("Cleaned shape:", cleaned_df.shape)

display(cleaned_df.head())
```

```python
print("TotalCharges datatype:", cleaned_df["TotalCharges"].dtype)

print("\nMissing values:")
display(cleaned_df.isnull().sum())

print("\nDuplicate customer IDs:",
      cleaned_df["customerID"].duplicated().sum())

print("\nOriginal data unchanged:",
      raw_df["TotalCharges"].dtype == object)
```

```python
missing_records = cleaned_df[
    cleaned_df["TotalCharges"].isnull()
]

display(missing_records[[
            "customerID",
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "Contract",
            "Churn"
        ]
        ]
    )
```

```python
print("Missing TotalCharges:", cleaned_df["TotalCharges"].isnull().sum())

print(
    "Tenure distribution among missing records:"
)
display(missing_records["tenure"].value_counts())
```

```python
import importlib
import customer_intelligence_churn_prediction_system.preprocessing.clean_data as cleaning

importlib.reload(cleaning)

clean_customer_data = cleaning.clean_customer_data

cleaned_df = clean_customer_data(raw_df)

print("Missing values after cleaning:")
display(cleaned_df.isnull().sum())

print(
    "Zero-tenure customers:",
    (cleaned_df["tenure"] == 0).sum()
)

print(
    "Zero-tenure customers with TotalCharges = 0:",
    (
        (cleaned_df["tenure"] == 0) &
        (cleaned_df["TotalCharges"] == 0)
    ).sum()
)
```

```python
from customer_intelligence_churn_prediction_system.features.build_features import (
    build_customer_features
)

featured_df = build_customer_features(cleaned_df)

print("Cleaned shape:", cleaned_df.shape)
print("Featured shape:", featured_df.shape)

display(
    featured_df[
        [
            "tenure",
            "TenureGroup",
            "ServiceCount",
            "Contract",
            "IsMonthToMonth",
            "TotalCharges",
            "AverageMonthlySpend"
        ]
    ].head(10)
)
```

```python
# Validate engineered features

print("Missing values:")
display(featured_df.isnull().sum())

print("\nTenure groups:")
display(featured_df["TenureGroup"].value_counts())

print("\nService count distribution:")
display(featured_df["ServiceCount"].value_counts().sort_index())

print("\nContract indicator:")
display(featured_df["IsMonthToMonth"].value_counts())

print("\nAverage monthly spend statistics:")
display(featured_df["AverageMonthlySpend"].describe())

print(
    "\nInfinite values:",
    np.isinf(featured_df["AverageMonthlySpend"]).sum()
)
```
