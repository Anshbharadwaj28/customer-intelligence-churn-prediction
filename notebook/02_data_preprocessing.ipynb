```python
from customer_intelligence_churn_prediction_system.features.build_features import (
    build_customer_features
)

featured_df = build_customer_features(cleaned_df)

print("Cleaned shape:", cleaned_df.shape)
print("Featured shape:", featured_df.shape)

display(
    featured_df[
        [
            "tenure",
            "TenureGroup",
            "ServiceCount",
            "Contract",
            "IsMonthToMonth",
            "TotalCharges",
            "AverageMonthlySpend"
        ]
    ].head(10)
)
```

```python
# Validate engineered features

print("Missing values:")
display(featured_df.isnull().sum())

print("\nTenure groups:")
display(featured_df["TenureGroup"].value_counts())

print("\nService count distribution:")
display(featured_df["ServiceCount"].value_counts().sort_index())

print("\nContract indicator:")
display(featured_df["IsMonthToMonth"].value_counts())

print("\nAverage monthly spend statistics:")
display(featured_df["AverageMonthlySpend"].describe())

print(
    "\nInfinite values:",
    np.isinf(featured_df["AverageMonthlySpend"]).sum()
)
```

```python
X = featured_df.drop(columns = ["customerID","Churn"])

y = featured_df["Churn"].map({
    "No":0,
    "Yes":1
})
print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
assert y.notna().all(), "Unexpected target values found."
assert set(y.unique()) == {0, 1}

print("Target encoding successful!")
```

```python
from customer_intelligence_churn_prediction_system.preprocessing.preprocess import (
    create_preprocessor
)

preprocessor = create_preprocessor(X)

print(preprocessor)
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(4) * 100)

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(4) * 100)
```

```python


X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training processed shape:", X_train_processed.shape)
print("Testing processed shape:", X_test_processed.shape)

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
import numpy as np

print("Training NaN values:", np.isnan(X_train_processed).sum())
print("Testing NaN values:", np.isnan(X_test_processed).sum())

print("Training infinite values:", np.isinf(X_train_processed).sum())
print("Testing infinite values:", np.isinf(X_test_processed).sum())
```
